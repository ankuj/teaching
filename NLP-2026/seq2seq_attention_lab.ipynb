```python
import numpy as np

def softmax(x, axis=-1):
    """Numerically stable softmax."""
    x = x - np.max(x, axis=axis, keepdims=True)
    exp_x = np.exp(x)
    return exp_x / np.sum(exp_x, axis=axis, keepdims=True)
```

## Part 2: The Bottleneck & Attention

Basic Seq2Seq compresses the entire input into a **single fixed-size context vector** — the bottleneck problem. Attention solves this by letting the decoder access **all encoder hidden states** via a weighted sum.

Attention equations (from lecture):
1. Alignment score: $e_{t,s} = \text{score}(h^{\text{dec}}_t, h^{\text{enc}}_s)$
2. Attention weights: $\alpha_{t,s} = \frac{\exp(e_{t,s})}{\sum_{s'} \exp(e_{t,s'})}$
3. Context vector: $c_t = \sum_s \alpha_{t,s} \, h^{\text{enc}}_s$
